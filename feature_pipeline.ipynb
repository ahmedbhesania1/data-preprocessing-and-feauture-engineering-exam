# OUTLIER DETECTION AND HANDLING

```python
# z score on amount 
mean=df['amount'].mean()
std=df['amount'].std()
df['amount_zscore']=(df['amount']-mean)/std

df=df[df['amount_zscore'].abs()< 3]
```

```python
# iqr on price
q1=df['price'].quantile(0.25)
q3=df['price'].quantile(0.75)
iqr=q3-q1
df=df[(df['price']>=q1-1.5*iqr) & (df['price']<=q3+1.5*iqr)]
```

# Handling Mixed & Date/Time Variables

```python
df['date']=pd.to_datetime(df['date'])
# days since last purchase
import pandas as pd


today = pd.Timestamp.today().normalize()

df['days_since_last_purchase'] = (today - df['date']).dt.days
```

```python
# splitting Trascation id and product id
df['t_id_letter'] = df['transaction_id'].str[0]
df['t_id_number'] = df['transaction_id'].str[1:].astype(int)
df['p_id_letter'] = df['product_id'].str[0]
df['p_id_number'] = df['product_id'].str[1:].astype(int)
df
```

```python
df.drop(columns=['transaction_id', 'product_id'], inplace=True)
```

# Encoding Categorical Data

```python
# label encoding
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['city_encoded']=le.fit_transform(df['city'])
```

```python
# one hot encoding on payment mode and product category
df=pd.get_dummies(df,columns=['payment_mode','category'],drop_first=True)
```

```python
# trying each scaler individually first
from sklearn.preprocessing import StandardScaler

std=StandardScaler()
df['age']=std.fit_transform(df[['age']])

```

```python
# minmax scaler on income
from sklearn.preprocessing import MinMaxScaler

mm=MinMaxScaler()
df['income_minmax']=mm.fit_transform(df[['income']])
```

```python
from sklearn.preprocessing import MaxAbsScaler
# maxabs scaler on amount
mabs=MaxAbsScaler()
df['amount_maxabs']=mabs.fit_transform(df[['amount']])
```

```python
from sklearn.preprocessing import RobustScaler
# robust scaler on price
rob=RobustScaler()
df['price_robust']=rob.fit_transform(df[['price']])
```

```python
from sklearn.preprocessing import Normalizer

# normalizer on stock
norm=Normalizer()
df['stock_norm']=norm.fit_transform(df[['stock']])
```

```python
df
```

```python
# now doing all of the above together using ColumnTransformer
from sklearn.compose import ColumnTransformer

ct=ColumnTransformer([
    ('standard',StandardScaler(),['age']),
    ('minmax',MinMaxScaler(),['income']),
    ('maxabs',MaxAbsScaler(),['amount']),
    ('robust',RobustScaler(),['price']),
    ('norm',Normalizer(),['stock'])
])

scaled_array=ct.fit_transform(df[['age','income','amount','price','stock']])
df_scaled=pd.DataFrame(scaled_array,columns=['age','income','amount','price','stock'],index=df.index)
df_scaled.head()
```

# Feature Construction & Transformation


```python
# total purchases per customer
df['total_purchases']=df.groupby('customer_id')['customer_id'].transform('count')
```

```python
df['purchase_per_day']=df['total_purchases']/df['days_since_last_purchase']
```

### FunctionTransformer (Log, Square Root, Reciprocal)

```python

from sklearn.preprocessing import FunctionTransformer
import numpy as np

log_t=FunctionTransformer(np.log1p)
df['amount_log']=log_t.fit_transform(df[['amount']])

sqrt_t=FunctionTransformer(np.sqrt)
df['price_sqrt']=sqrt_t.fit_transform(df[['price']])

recip_t=FunctionTransformer(lambda x:1/(x+1))
df['stock_recip']=recip_t.fit_transform(df[['stock']])
```

```python
from sklearn.preprocessing import PowerTransformer
pt_yj=PowerTransformer(method='yeo-johnson')
df['amount_yeojohnson']=pt_yj.fit_transform(df[['amount']])

pt_bc=PowerTransformer(method='box-cox')
df['price_boxcox']=pt_bc.fit_transform(df[['price']])
```

### Equal-width (quantile) binning on income.


```python
df['income_bin']=pd.qcut(df['income'],q=4,labels=['q1','q2','q3','q4'])
```

### Binary conversion of purchase frequency (frequent_buyer = 1 if > threshold).


```python

threshold=df['total_purchases'].median()
df['frequent_buyer']=(df['total_purchases']>threshold).astype(int)
```

- What techniques were used?

Preprocessing used label encoding (city, gender), one-hot encoding (payment mode, category) and ordinal encoding (income groups, since the dataset didn’t have a direct ordinal column). For scaling, StandardScaler, MinMaxScaler, RobustScaler and Normalizer were tried, and also applied together using ColumnTransformer.

- What were the biggest issues with the raw data?

The biggest issue in the raw data was mixed columns like transaction_id/product_id, along with right-skewed amount/price columns, which were handled using log and Yeo-Johnson transforms. New features such as total_purchases, purchase_per_day and frequent_buyer were added to the final feature set.

- Which imputation and scaling worked best?

There were no missing values in the dataset, so no imputation techniques needed to be applied.
